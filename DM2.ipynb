```python
# Task 1: Apply Indexing and Slicing

import pandas as pd
import numpy as np
```

```python
# Load Dataset Using Pandas

df = pd.read_csv("/content/Retail_Dataset_Modified.csv")

print("Dataset Loaded Successfully")
```

```python
# Create NumPy Array from Revenue Column

revenue_array = np.array(df["Revenue"])

print(revenue_array)
```

```python
# Access the First Revenue Value Using Indexing

print("First Revenue Value:")
print(revenue_array[0])
```

```python
# Access the Last Revenue Value

print("Last Revenue Value:")
print(revenue_array[-1])
```

```python
# Slice the First Five Revenue Records

print("First Five Revenue Values:")
print(revenue_array[0:5])
```

```python
print("Revenue Values from Index 10 to 20:")
print(revenue_array[10:21])
```

```python
sales_array = np.array(df[["Units_Sold", "Profit_Margin"]])

print(sales_array)
```

```python
print("First Row:")
print(sales_array[0])
```

```python
print("Profit Margin Column:")
print(sales_array[:,1])
```

```python
# Task 2: Perform Operations

# Display the First 10 Revenue Values

first_ten = revenue_array[0:10]
print(first_ten)
```

```python
# Find Total Revenue of First 10 Records

print("Total Revenue:")
print(np.sum(first_ten))
```

```python
# Find Highest Revenue from First 10 Records

print("Maximum Revenue:")
print(np.max(first_ten))
```

```python
# Find Lowest Revenue from First 10 Records

print("Lowest Revenue:")
print(np.min(first_ten))
```
