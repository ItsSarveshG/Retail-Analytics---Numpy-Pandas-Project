```python
# Task 2: Handle missing values
# Check Missing Values

print("Missing Values:")
print(df.isnull().sum())
```

```python
# Display Columns Having Missing Values

missing_cols = df.isnull().sum()
print(missing_cols[missing_cols > 0])
```

```python
# Fill Missing Values in City Using Mode

df['City'].fillna(df['City'].mode()[0], inplace = False)
```

```python
# Fill Missing Values in Product Name Using Mode

df['Product_Name'].fillna(df['Product_Name'].mode()[0], inplace=True)
```

```python
# Fill Missing Values in Quantity Available Using Mean

df['Quantity_Available'].fillna(df['Quantity_Available'].mean(), inplace=True)
```

```python
# Fill Missing Values in Units Sold Using Mean

df['Units_Sold'].fillna(df['Units_Sold'].mean(),inplace=True)
```

```python
# Fill Missing Values in Revenue Using Mean

df['Revenue'].fillna(df['Revenue'].mean(), inplace=True)
```

```python
# Fill Missing Values in Supplier Using Mode

df['Supplier'].fillna(df['Supplier'].mode()[0],inplace=True)
```

```python
# Fill Missing Values in Customer Satisfaction Using Mean

df['Customer_Satisfaction'].fillna(df['Customer_Satisfaction'].mean(),inplace=True)
```

```python
# Fill Missing Values in Delivery Time Using Mean

df['Delivery_Time'].fillna(df['Delivery_Time'].mean(),inplace=True)
```

```python
# Fill Missing Values in Shipping Cost Using Mean

df['Shipping_Cost'].fillna(df['Shipping_Cost'].mean(),inplace=True)
```

```python
# Verify Missing Values After Cleaning

print(df.isnull().sum())
```
