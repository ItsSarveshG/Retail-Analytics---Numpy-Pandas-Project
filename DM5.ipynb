```python
# Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Load Dataset Using Pandas

df = pd.read_csv("/content/Retail_Dataset_Cleaned_1509.csv")

print("Dataset Loaded Successfully")
```

```python
# Display First Five Records

df.head()
```

```python
# Check Dataset Information

df.info()
```

```python
# Create Profit Amount Column

df["Profit_Amount"] = (df["Revenue"] * df["Profit_Margin"] / 100)

df[["Revenue","Profit_Margin","Profit_Amount"]].sample(10)
```

```python
# Create Stock Difference Column

df["Stock_Difference"] = (df["Quantity_Available"] - df["Units_Sold"])

df[["Quantity_Available","Units_Sold","Stock_Difference"]].head()
```

```python
# Create Inventory Status Column

df["Inventory_Status"] = np.where(df["Quantity_Available"] < 100, "Low Stock",np.where(df["Quantity_Available"] <= 200, "Medium Stock","High Stock"))

df[["Quantity_Available","Inventory_Status"]].sample(3)
```

```python
# Create Inventory Status Column

df["Inventory_Status"] = np.where(
    df["Quantity_Available"] < 100,
    "Low Stock",
    np.where(
        df["Quantity_Available"] <= 200,
        "Medium Stock",
        "High Stock"
    )
)

df[["Quantity_Available",
    "Inventory_Status"]].sample(5)
```

```python
# Create Revenue Per Unit Column

df["Revenue_Per_Unit"] = (
    df["Revenue"] /
    df["Units_Sold"]
)

df[["Revenue",
    "Units_Sold",
    "Revenue_Per_Unit"]].sample(5)
```

```python
# Verify Newly Created Columns

print(df[[
    "Profit_Amount",
    "Stock_Difference",
    "Inventory_Status",
    "Revenue_Per_Unit"
]].head())
```

```python
# Task 2: Transform Data

# Transform Revenue into Revenue Category

df["Revenue_Category"] = np.where(
    df["Revenue"] < 5000,
    "Low Revenue",
    np.where(
        df["Revenue"] < 10000,
        "Medium Revenue",
        "High Revenue"
    )
)

df[["Revenue", "Revenue_Category"]].head()
```

```python
# Transform Customer Satisfaction Ratings

df["Customer_Feedback"] = np.where(
    df["Customer_Satisfaction"] >= 4,
    "Satisfied",
    "Needs Improvement"
)

df[[
    "Customer_Satisfaction",
    "Customer_Feedback"
]].head()
```

```python
# Create Discount Category

df["Discount_Category"] = np.where(
    df["Max_Discount_Offered"] < 10,
    "Low Discount",
    np.where(
        df["Max_Discount_Offered"] < 20,
        "Medium Discount",
        "High Discount"
    )
)

df[["Max_Discount_Offered", "Discount_Category"]].sample(5)
```

```python
# Create Delivery Speed Category

df["Delivery_Speed"] = np.where(df["Delivery_Time"] <= 5,"Fast Delivery", "Standard Delivery")

df[["Delivery_Time","Delivery_Speed"]].sample(10)
```

```python
# Display Transformed Columns

print(df[[
    "Revenue_Category",
    "Customer_Feedback",
    "Discount_Category",
    "Delivery_Speed"
]].head())
```
