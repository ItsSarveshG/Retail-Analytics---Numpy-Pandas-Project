```python
# Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Load Dataset Using Pandas

df = pd.read_csv("/content/Retail_Dataset_Cleaned_1509_lab6.csv")

print("Dataset Loaded Successfully")
```

```python
# Display First Five Records

df.head()
```

```python
# Check Dataset Information

df.info()
```

```python
# Calculate Mean Revenue

mean_revenue = df["Revenue"].mean()

print("Mean Revenue:", mean_revenue)
```

```python
# Calculate Median Revenue

median_revenue = df["Revenue"].median()

print("Median Revenue:", median_revenue)
```

```python
# Compare Mean and Median

comparison = pd.DataFrame({
    "Statistic":["Mean Revenue", "Median Revenue"],
    "Value": [mean_revenue, median_revenue]
})

comparison
```
